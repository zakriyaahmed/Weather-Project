```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('weather.csv') 
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["WindDir9am"].ffill(inplace=True)
```

```python
df["Sunshine"].fillna(df["Sunshine"].mean(),inplace=True)
```

```python
df["WindGustDir"].fillna(df["WindGustDir"].mode()[0],inplace=True)  
```

```python
df["WindSpeed9am"].fillna(df["WindSpeed9am"].mean(),inplace=True)
```

```python
df["WindGustSpeed"].fillna(df["WindGustSpeed"].mean(),inplace=True)
```

```python
df["WindDir3pm"].fillna(df["WindDir3pm"].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(x=df['RainToday'])
plt.title("Rain Today Count")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['MinTemp'], bins=20,  color='blue')
plt.title("Distribution of Min Temperature")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Heatmap of Feature Correlations",color="red").set_fontsize(40)
plt.show()
```

```python
plt.figure(figsize=(6, 6))
df['RainToday'].value_counts().plot.pie(autopct='%1.1f%%', colors=['skyblue', 'lightcoral'])
plt.title("Rain Today Distribution")
plt.show()
```

```python
sns.countplot(x=df["RainToday"],data=df,hue="RainToday")
plt.show()
```

```python
from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(df['RainToday'], df['RainTomorrow'])
chi2, p, dof, expected = chi2_contingency(contingency_table)
plt.figure(figsize=(6,4))
sns.heatmap(contingency_table, annot=True, cmap="Blues", fmt='d')
plt.title("Chi-Square Contingency Table Heatmap").set_fontsize(40)
plt.show()
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
```
